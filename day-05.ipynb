```python
number = [1,5,6,8]
inputNum = int(input("Enter number: "))
if inputNum in number:
    idx = number.index(inputNum)
    print('index of ', inputNum, 'is', idx)
    number.remove(inputNum)
else:
    print("Number not exist")
print(number)
```
