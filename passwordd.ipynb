```python
password="open123"
attempts=3
for i in range (1,6):
    user_input=str(input("Enter a password:"))
    if user_input==password:
         print("Access granted")
         break
    elif user_input=='forgot':
        continue
    else:
        attempts-=1
        print(f"wrong password! {attempts} attempts left")
```
